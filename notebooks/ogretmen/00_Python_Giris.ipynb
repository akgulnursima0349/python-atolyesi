## 3️⃣ Python bir Hesap Makinesi 🧮

| İşlem | Sembol | Örnek | Sonuç |
|-------|--------|-------|-------|
| Toplama | `+` | `7 + 3` | `10` |
| Çıkarma | `-` | `7 - 3` | `4` |
| Çarpma | `*` | `7 * 3` | `21` |
| Bölme | `/` | `7 / 2` | `3.5` |
| Tam bölme | `//` | `7 // 2` | `3` (küsuratı atar) |
| Kalan (mod) | `%` | `7 % 2` | `1` (bölümden kalan) |
| Üs alma | `**` | `2 ** 3` | `8` (2×2×2) |

````python
print(7 + 3)
print(7 - 3)
print(7 * 3)
print(7 / 2)    # normal bölme → ondalıklı sonuç verir
print(7 // 2)   # tam bölme → sadece tam kısmı verir
print(7 % 2)    # mod (kalan): 7'yi 2'ye bölünce kalan 1
print(2 ** 10)  # 2'nin 10. kuvveti
````

````python
# Gerçek hayattan bir örnek: Robot kursu için kaç tane robot seti lazım?
ogrenci_sayisi = 23
kisi_basina_grup = 4

tam_grup = ogrenci_sayisi // kisi_basina_grup   # kaç tane DOLU grup var?
artan = ogrenci_sayisi % kisi_basina_grup       # gruba girmeyen kaç kişi kaldı?

print("Dolu grup sayısı:", tam_grup)
print("Artan öğrenci:", artan)

# Python matematikteki işlem önceliğine uyar: önce çarpma/bölme, sonra toplama/çıkarma
print(2 + 3 * 4)     # 14 (önce 3*4)
print((2 + 3) * 4)   # 20 (önce parantez)
````

> 💡 `%` (mod) işlemi çok işe yarar! Bir sayının **çift mi tek mi** olduğunu anlamak için `sayi % 2` hesaplarız: sonuç `0` ise çift, `1` ise tektir.

## 5️⃣ Metinlerle Oynamak (*String*) 🔤

Metinler harflerin yan yana dizilmesiyle oluşur. Python'da metinlerle çok güzel oyunlar oynayabiliriz.

````python
mesaj = "Robotik Akademi"

print(len(mesaj))       # len (length = uzunluk): metinde kaç karakter var? (boşluk da sayılır!)
print(mesaj.upper())    # upper (yukarı/büyük): hepsini BÜYÜK harf yapar
print(mesaj.lower())    # lower (aşağı/küçük): hepsini küçük harf yapar
print(mesaj[0])         # [0] → ilk harf. DİKKAT: Python saymaya 0'dan başlar!
print(mesaj[-1])        # [-1] → son harf (eksi sayılar sondan saymak demek)
print(mesaj[0:7])       # [0:7] → 0. harften 7. harfe kadar (7 dahil değil) → "Robotik"
````

````python
# Metinleri '+' ile birleştirebilir, '*' ile çoğaltabiliriz
print("Pi" + "ton")     # birleştirme → "Piton"
print("Ha" * 5)         # 5 kere tekrar → "HaHaHaHaHa"
print("=" * 30)         # süslü bir çizgi yapmak için çok kullanışlı!
````

### ✨ f-string: En kullanışlı metin yazma yöntemi
Metnin başına `f` harfi koyarsak (*f* = *format*, yani biçimlendirme), süslü parantez `{ }` içine değişken yazabiliriz.
Python o değişkenin değerini metnin içine yerleştirir.

````python
isim = "Ece"
puan = 95

# Eski yöntem (virgüllerle) biraz karışık:
print("Oyuncu", isim, "toplam", puan, "puan aldı.")

# f-string ile çok daha okunaklı:
print(f"Oyuncu {isim} toplam {puan} puan aldı.")

# Süslü parantez içinde hesap bile yapabiliriz:
print(f"{isim} 5 puan daha alırsa {puan + 5} puanı olur!")

# Ondalıklı sayıları güzel göstermek: ':.2f' → virgülden sonra 2 basamak göster
pi = 3.14159265
print(f"Pi sayısı yaklaşık {pi:.2f}")
````

## 6️⃣ Bilgisayara Karar Verdirmek: `if` 🚦

`if` (İngilizce "eğer" demek) ile bilgisayara **"Eğer şu doğruysa, bunu yap"** diyebiliriz.

```
if  koşul:            → eğer koşul doğruysa
    bunu yap          → (4 boşluk içeride!)
elif başka_koşul:     → değilse ve eğer bu doğruysa   (elif = else if)
    şunu yap
else:                 → hiçbiri doğru değilse   (else = değilse)
    onu yap
```

> ⚠️ Satırın sonundaki **iki nokta** `:` ve alt satırdaki **girinti** (*indentation* — satır başındaki 4 boşluk) çok önemli! Python hangi komutların `if`'e ait olduğunu girintiye bakarak anlar.

### Karşılaştırma işaretleri
| İşaret | Anlamı | Örnek | Sonuç |
|--------|--------|-------|-------|
| `==` | eşit mi? | `5 == 5` | `True` |
| `!=` | eşit değil mi? | `5 != 3` | `True` |
| `>` | büyük mü? | `5 > 3` | `True` |
| `<` | küçük mü? | `5 < 3` | `False` |
| `>=` | büyük veya eşit mi? | `5 >= 5` | `True` |
| `<=` | küçük veya eşit mi? | `2 <= 1` | `False` |

> `=` (tek) → kutuya değer koyar. `==` (çift) → iki şey eşit mi diye **sorar**. Bunları karıştırmamak çok önemli!

````python
sicaklik = 28

if sicaklik > 30:
    print("Çok sıcak! 🥵 Robotun motorları ısınabilir.")
elif sicaklik > 15:
    print("Hava güzel! 😎 Bahçede robot yarışı yapabiliriz.")
else:
    print("Hava soğuk! 🥶 Sınıfta çalışalım.")

# 'sicaklik' değerini 35 ya da 5 yapıp hücreyi tekrar çalıştır, ne değişiyor?
````

### `and` (ve), `or` (veya), `not` (değil)
Birden fazla koşulu birleştirebiliriz:
- `and` (ve): **iki** koşul da doğruysa → `True`
- `or` (veya): koşullardan **en az biri** doğruysa → `True`
- `not` (değil): doğruyu yanlış, yanlışı doğru yapar

````python
pil = 45            # robotun pil yüzdesi
engel_var = False   # önünde engel var mı?

if pil > 20 and not engel_var:
    print("🤖 Robot ilerliyor...")
elif pil <= 20:
    print("🔋 Pil azaldı, şarja gidiyorum!")
else:
    print("🚧 Önümde engel var, dönüyorum!")
````

````python
# Sayı tek mi çift mi? (Hatırla: % → bölümden kalan)
sayi = int(input("Bir sayı yaz: "))

if sayi % 2 == 0:
    print(f"{sayi} bir ÇİFT sayıdır.")
else:
    print(f"{sayi} bir TEK sayıdır.")
````

## 7️⃣ Listeler: Bilgi Trenleri 🚃🚃🚃

**Liste** (*list*), birden çok bilgiyi **sırayla** tutan bir yapıdır. Köşeli parantez `[ ]` ile yazılır, elemanlar virgülle ayrılır.
Her elemanın bir **sıra numarası** (*index* — indeks, yani konum numarası) vardır ve numaralar **0'dan başlar**.

```
robotlar = [ "Robi" , "Mbot" , "Lego" ]
```

````python
robotlar = ["Robi", "Mbot", "Lego", "Arduino"]

print(robotlar)          # tüm liste
print(robotlar[0])       # ilk eleman (0. sıra)
print(robotlar[-1])      # son eleman
print(len(robotlar))     # len (uzunluk): listede kaç eleman var?
````

````python
# Listeyi değiştirmek
robotlar.append("Micro:bit")   # append (sonuna ekle): listenin sonuna yeni eleman ekler
print("Ekledikten sonra:", robotlar)

robotlar.remove("Lego")        # remove (kaldır): bu elemanı listeden siler
print("Sildikten sonra:", robotlar)

robotlar[0] = "Süper Robi"     # 0. sıradaki elemanı değiştir
print("Değiştirdikten sonra:", robotlar)

print("Mbot listede var mı?", "Mbot" in robotlar)   # 'in' (içinde): eleman listede var mı?
````

## 8️⃣ Döngüler: Tekrar Tekrar Yap! 🔁

Bir işi 100 kere yaptırmak için 100 satır yazmak yerine **döngü** (*loop*) kullanırız.

### `for` döngüsü — "her biri için"
`for` (İngilizce "için") döngüsü, bir listedeki **her eleman için** aynı işi yapar.

````python
meyveler = ["elma", "muz", "çilek"]

for meyve in meyveler:          # meyveler listesindeki HER BİR meyve için...
    print(f"Ben {meyve} severim! 😋")   # ...bu satırı çalıştır (girintiye dikkat!)

print("Döngü bitti.")           # bu satır girintili değil → döngünün DIŞINDA, 1 kere çalışır
````

### `range` — sayı üreten makine
`range` (aralık) belirli aralıkta sayılar üretir:
- `range(5)` → 0, 1, 2, 3, 4  (5 dahil değil!)
- `range(1, 6)` → 1, 2, 3, 4, 5
- `range(0, 20, 5)` → 0, 5, 10, 15 (5'er 5'er atla)

````python
# 1'den 10'a kadar sayıların karelerini yazdıralım
for i in range(1, 11):
    print(f"{i} x {i} = {i * i}")
````

````python
# Çarpım tablosu: İÇ İÇE döngü (bir döngünün içinde başka bir döngü)
for satir in range(1, 6):
    for sutun in range(1, 6):
        print(f"{satir * sutun:4}", end="")   # :4 → sayıyı 4 karakterlik alana yaz (hizalı dursun)
                                              # end="" → alt satıra geçme, yan yana yaz
    print()   # her satırın sonunda alt satıra geç
````

### `while` döngüsü — "olduğu sürece"
`while` (İngilizce "-iken / olduğu sürece") döngüsü, koşul **doğru olduğu sürece** tekrar eder.

> ⚠️ Koşul hiç yanlış olmazsa döngü **sonsuza kadar** döner! Böyle olursa hücrenin solundaki ⏹️ (durdur) düğmesine bas.

````python
# Robotun pili bitene kadar ilerlemesi
pil = 100
adim = 0

while pil > 0:            # pil 0'dan büyük OLDUĞU SÜRECE...
    adim += 1             # bir adım at
    pil -= 15             # her adımda pil 15 azalır
    print(f"Adım {adim}: pil %{max(pil, 0)}")

print(f"🔋 Pil bitti! Robot toplam {adim} adım attı.")
````

````python
# break (kır/çık): döngüden ANINDA çıkar
# Listede ilk negatif sayıyı bulalım
sensor_okumalari = [12, 8, 15, -3, 20, 7]

for deger in sensor_okumalari:
    if deger < 0:
        print(f"⚠️ Hatalı okuma bulundu: {deger}. Durduruluyor!")
        break
    print(f"Okuma tamam: {deger}")
````

## 9️⃣ Fonksiyonlar: Kendi Komutlarını Yaz! 🛠️

**Fonksiyon** (*function*), bir işi yapan kod parçasına **isim** vermemizdir. Bir kere yazarız, istediğimiz kadar kullanırız.
`print`, `len`, `max` aslında Python'un hazır fonksiyonlarıdır. Şimdi kendimizinkini yazacağız!

```python
def fonksiyon_adi(parametre):     ← def (define = tanımla)
    yapılacak işler
    return sonuç                   ← return (geri döndür) = sonucu dışarı ver
```
- **Parametre** (*parameter*): Fonksiyona dışarıdan verdiğimiz bilgi (malzeme).
- **return** (geri döndür): Fonksiyonun ürettiği sonucu dışarı verir.

````python
def selamla(isim):                       # 'selamla' adında, 'isim' parametresi alan bir fonksiyon tanımladık
    print(f"Merhaba {isim}! Robotik Akademi'ye hoş geldin 🤖")

# Fonksiyonu ÇAĞIRMAK (kullanmak):
selamla("Kerem")
selamla("Elif")
selamla("Yusuf")
````

````python
def dikdortgen_alani(en, boy):     # iki parametre alan fonksiyon
    alan = en * boy
    return alan                    # sonucu dışarı ver

sonuc = dikdortgen_alani(5, 3)     # fonksiyonun döndürdüğü değer 'sonuc' kutusuna girer
print("Alan:", sonuc)
print("Sınıfın alanı:", dikdortgen_alani(8, 6), "metrekare")
````

# 🏆 Mini Proje 1: Sayı Tahmin Oyunu 🎲

Öğrendiğimiz **her şeyi** bir araya getiriyoruz: değişkenler, `input`, `if`, `while`, f-string…

Bilgisayar 1 ile 100 arasında gizli bir sayı tutacak. Sen tahmin edeceksin, bilgisayar sana "daha büyük" ya da "daha küçük" diyecek. **7 hakkın var!**

> 📦 `import random` → Python'un hazır **kütüphanesinden** (*library* — başkalarının yazıp paylaştığı hazır kod kutusu) `random` (rastgele) modülünü getirir.

````python
import random                         # random (rastgele): rastgele sayı üretmek için hazır modül

gizli_sayi = random.randint(1, 100)   # randint (random integer): 1 ile 100 arasında rastgele tam sayı
hak = 7
tahmin_sayisi = 0
kazandi = False

print("🎲 1 ile 100 arasında bir sayı tuttum. Bakalım bulabilecek misin?")
print(f"Toplam {hak} hakkın var.\n")    # \n → yeni satır karakteri (alt satıra geç)

while tahmin_sayisi < hak:
    tahmin = int(input(f"Tahmin #{tahmin_sayisi + 1}: "))
    tahmin_sayisi += 1

    if tahmin == gizli_sayi:
        kazandi = True
        break                          # bildin → döngüden çık
    elif tahmin < gizli_sayi:
        print("   ⬆️ Daha BÜYÜK bir sayı söyle!")
    else:
        print("   ⬇️ Daha KÜÇÜK bir sayı söyle!")

if kazandi:
    print(f"\n🎉 TEBRİKLER! {tahmin_sayisi} tahminde buldun!")
else:
    print(f"\n😢 Hakların bitti. Gizli sayı {gizli_sayi} idi.")
````

> 🧠 **Düşünelim:** Bu oyunu her seferinde 7 tahminde kazanmanın bir yolu var! Her seferinde kalan aralığın **tam ortasını** söylersen (1-100 → 50, sonra 25 ya da 75…) aralık her seferinde yarıya iner. Buna bilgisayar biliminde **ikili arama** (*binary search*) denir. 100 → 50 → 25 → 13 → 7 → 4 → 2 → 1. Yani 7 tahmin her zaman yeter! 🤯

# 🏆 Mini Proje 2: Robot Simülatörü 🤖

Robotik dersinde robotlarımıza komut veriyoruz. Şimdi bilgisayarda **sanal bir robot** yapalım!

Robotumuz 10×10'luk bir ızgarada (*grid*) yaşıyor. Ona bir **komut listesi** vereceğiz:
- `"ileri"` → baktığı yöne 1 adım git
- `"sag"` → 90° sağa dön
- `"sol"` → 90° sola dön

Robot ızgaradan dışarı çıkmaya çalışırsa duvara çarpar ve o adımı atlar.

````python
def robotu_calistir(komutlar, boyut=10):
    x, y = 0, 0                    # robot sol üst köşeden (0, 0) başlar
    yon = "dogu"                   # başlangıçta sağa (doğuya) bakıyor
    yonler = ["kuzey", "dogu", "guney", "bati"]   # saat yönünde sıralı yönler
    hareket = {                    # her yön için x ve y'nin nasıl değişeceği
        "kuzey": (0, -1), "dogu": (1, 0), "guney": (0, 1), "bati": (-1, 0)
    }
    iz = [(x, y)]                  # robotun geçtiği tüm kareleri kaydedelim

    for komut in komutlar:
        if komut == "sag":
            yon = yonler[(yonler.index(yon) + 1) % 4]   # listede bir sonraki yön (sonda başa dön → % 4)
        elif komut == "sol":
            yon = yonler[(yonler.index(yon) - 1) % 4]   # listede bir önceki yön
        elif komut == "ileri":
            dx, dy = hareket[yon]
            yeni_x, yeni_y = x + dx, y + dy
            if 0 <= yeni_x < boyut and 0 <= yeni_y < boyut:   # ızgaranın içinde mi?
                x, y = yeni_x, yeni_y
                iz.append((x, y))
            else:
                print(f"💥 Duvara çarptım! ({x},{y}) noktasında {yon} yönüne gidemiyorum.")
        else:
            print(f"❓ '{komut}' komutunu anlamadım.")
    return iz, yon


def haritayi_ciz(iz, yon, boyut=10):
    oklar = {"kuzey": "⬆️", "dogu": "➡️", "guney": "⬇️", "bati": "⬅️"}
    for y in range(boyut):
        satir = ""
        for x in range(boyut):
            if (x, y) == iz[-1]:
                satir += oklar[yon]    # robotun şu anki yeri ve yönü
            elif (x, y) == iz[0]:
                satir += "🏠"           # başlangıç noktası
            elif (x, y) in iz:
                satir += "🟩"           # robotun geçtiği yerler
            else:
                satir += "⬜"           # boş kare
        print(satir)


komutlar = ["ileri", "ileri", "ileri", "sag", "ileri", "ileri", "sol", "ileri", "ileri", "sag", "ileri", "ileri", "ileri"]
iz, son_yon = robotu_calistir(komutlar)
haritayi_ciz(iz, son_yon)
yon_adi = {"kuzey": "kuzey", "dogu": "doğu", "guney": "güney", "bati": "batı"}   # ekranda Türkçe karakterlerle gösterelim
print(f"\n📍 Robot şu an {iz[-1]} noktasında ve {yon_adi[son_yon]} yönüne bakıyor. Toplam {len(iz) - 1} adım attı.")
````
